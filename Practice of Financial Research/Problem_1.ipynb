# Problem 1: Optimal Investment Thresholds with Downward Jumps

This notebook solves for the economically relevant root $\beta>1$, calculates $V^*=\frac{\beta}{\beta-1}F$, and exports both a LaTeX table fragment and its wrapper.

```python
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.optimize import brentq
from IPython.display import display

F = 1.0
sigma = 0.20
alpha = 0.06
mu = 0.10
r = 0.04

lambda_values = [0.1, 0.2]
phi_values = [0.0, 0.4, 0.8, 1.0]
```

## Solve the fundamental equation

```python
def fundamental_equation(epsilon, jump_intensity, jump_size):
    return (
        0.5 * sigma**2 * epsilon * (epsilon - 1)
        + (r + alpha - mu) * epsilon
        + jump_intensity * ((1 - jump_size)**epsilon - 1)
        - r
    )


rows = []
for jump_intensity in lambda_values:
    for jump_size in phi_values:
        epsilon = brentq(
            fundamental_equation,
            1.0 + 1e-8,
            100.0,
            args=(jump_intensity, jump_size),
        )
        v_star = (epsilon / (epsilon - 1)) * F
        rows.append([jump_intensity, jump_size, epsilon, v_star])

results = pd.DataFrame(rows, columns=['Lambda', 'Phi', 'Epsilon', 'V_star'])
display(results.round(4))
```

## Export the LaTeX fragment and wrapper

```python
tables_dir = Path('Tables')
tables_dir.mkdir(parents=True, exist_ok=True)

fragment_path = tables_dir / 'Problem1_NumericalResults.tex'

with fragment_path.open('w', encoding='utf-8') as latex_file:
    latex_file.write(r'$\lambda$ & $\phi$ & $\varepsilon$ & $V^{*}$ \\' + '\n')
    latex_file.write(r'\midrule' + '\n')

    for row in results.itertuples(index=False):
        latex_file.write(
            f'{row.Lambda:.1f} & {row.Phi:.1f} & '
            f'{row.Epsilon:.4f} & {row.V_star:.4f} '
            + r'\\' + '\n'
        )

    latex_file.write(r'\bottomrule' + '\n')

print(f'Table fragment created: {fragment_path.resolve()}')
```
